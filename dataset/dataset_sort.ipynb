```python
import dask.dataframe as dd
```

```python

df = dd.read_csv("trending_yt_videos_113_countries.csv", blocksize=None, quotechar='"', doublequote=True, on_bad_lines='skip', dtype=str)
```

```python
df = df.head(50000, compute=False)
```

```python
df = df.drop(columns=["daily_rank", "daily_movement", "weekly_movement", "country", "description", "thumbnail_url", "video_tags", "kind"])
```

```python
df["snapshot_date"] = dd.to_datetime(df["snapshot_date"])
```

```python
df_partitioned = df.shuffle(on="video_id", npartitions=5, shuffle="tasks")
```

```python
df_sorted = df_partitioned.map_partitions(
    lambda partitioned_df: partitioned_df.sort_values(by="snapshot_date", ascending=True)
)
```

```python
df_sorted.to_csv("./dataset_partition-*.csv", index=False)
```
